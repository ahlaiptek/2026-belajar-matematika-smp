```python
# ============================================================
# BILANGAN BERPANGKAT
# Materi: Pangkat Negatif
# ============================================================

print("=" * 60)
print("              BILANGAN BERPANGKAT")
print("                  PANGKAT NEGATIF")
print("=" * 60)


# ------------------------------------------------------------
# 1. Pengertian Pangkat Negatif
# ------------------------------------------------------------

print("\n1. PENGERTIAN PANGKAT NEGATIF")
print("-" * 60)

print("Pangkat negatif berarti kebalikan dari pangkat positif.")
print()
print("Rumus:")
print("a^(-n) = 1 / a^n")
print()


# ------------------------------------------------------------
# 2. Contoh 5 pangkat -1
# ------------------------------------------------------------

print("2. CONTOH: 5 pangkat -1")
print("-" * 60)

basis = 5
pangkat = -1

hasil = basis ** pangkat

print("Basis   :", basis)
print("Pangkat :", pangkat)
print("Hasil   :", hasil)

print()
print("Secara matematika:")
print("5^(-1) = 1 / 5^1")
print("       = 1 / 5")
print("       = 0.2")


# ------------------------------------------------------------
# 3. Contoh 1 pangkat -2
# ------------------------------------------------------------

print("\n3. CONTOH: 1 pangkat -2")
print("-" * 60)

basis = 1
pangkat = -2

hasil = basis ** pangkat

print("Basis   :", basis)
print("Pangkat :", pangkat)
print("Hasil   :", hasil)

print()
print("Secara matematika:")
print("1^(-2) = 1 / 1^2")
print("       = 1 / 1")
print("       = 1")


# ------------------------------------------------------------
# 4. Contoh 1 pangkat -3
# ------------------------------------------------------------

print("\n4. CONTOH: 1 pangkat -3")
print("-" * 60)

basis = 1
pangkat = -3

hasil = basis ** pangkat

print("Basis   :", basis)
print("Pangkat :", pangkat)
print("Hasil   :", hasil)

print()
print("Secara matematika:")
print("1^(-3) = 1 / 1^3")
print("       = 1 / 1")
print("       = 1")


# ------------------------------------------------------------
# 5. Contoh 2 pangkat -4
# ------------------------------------------------------------

print("\n5. CONTOH: 2 pangkat -4")
print("-" * 60)

basis = 2
pangkat = -4

hasil = basis ** pangkat

print("Basis   :", basis)
print("Pangkat :", pangkat)
print("Hasil   :", hasil)

print()
print("Secara matematika:")
print("2^(-4) = 1 / 2^4")
print("       = 1 / 16")
print("       = 0.0625")


# ------------------------------------------------------------
# 6. Membandingkan pangkat positif dan negatif
# ------------------------------------------------------------

print("\n6. PERBANDINGAN PANGKAT POSITIF DAN NEGATIF")
print("-" * 60)

basis = 2

print("2^4  =", 2 ** 4)
print("2^3  =", 2 ** 3)
print("2^2  =", 2 ** 2)
print("2^1  =", 2 ** 1)
print("2^0  =", 2 ** 0)
print("2^-1 =", 2 ** -1)
print("2^-2 =", 2 ** -2)
print("2^-3 =", 2 ** -3)
print("2^-4 =", 2 ** -4)


# ------------------------------------------------------------
# 7. Membuat pola pangkat negatif
# ------------------------------------------------------------

print("\n7. POLA PANGKAT NEGATIF")
print("-" * 60)

basis = 2

for pangkat in range(-1, -6, -1):
    hasil = basis ** pangkat

    print(
        f"{basis}^{pangkat:>2} = {hasil}"
    )


# ------------------------------------------------------------
# 8. Menampilkan bentuk pecahan
# ------------------------------------------------------------

print("\n8. BENTUK PECAHAN")
print("-" * 60)

basis = 2

for pangkat in range(1, 6):
    hasil = basis ** pangkat

    print(
        f"{basis}^-{pangkat} = 1/{basis}^{pangkat} = 1/{hasil}"
    )


# ------------------------------------------------------------
# 9. Kesimpulan
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("KESIMPULAN")
print("=" * 60)

print("""
Pangkat negatif dapat diubah menjadi pecahan.

Rumus:

a^(-n) = 1 / a^n

Contoh:

5^(-1) = 1/5

2^(-2) = 1/2²
       = 1/4

2^(-4) = 1/2⁴
       = 1/16
""")
```
