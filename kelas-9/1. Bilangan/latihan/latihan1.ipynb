```python
# Aku ingin coba latihan bilangan bentuk akar
"""
Misal:
    3^2 = 9
    100^2 = 10_000
Maka akar dari 9 adalah 3, dan akar dari 10_000 adalah 100
"""
import math
print("Akar dari 9 adalah", math.sqrt(9))
print("Akar dari 10_000 adalah", math.sqrt(10_000))
print()
print("Pangkat 2 dari 3 adalah", math.pow(3, 2))
print("Pangkat 2 dari 100 adalah", math.pow(100, 2))
```
