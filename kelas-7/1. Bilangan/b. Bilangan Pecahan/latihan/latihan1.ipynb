Operasi pecahan


```python
from fractions import Fraction

a = Fraction(1, 2)
b = Fraction(3, 4)

print("Bilangan pecahan a:", a)
print("Bilangan pecahan b:", b)
print()

print("Hasil penjumlahan a + b:", a + b)
print("Hasil pengurangan a - b:", a - b)
print()

a = Fraction(2, 4)
b = Fraction(6, 8)
print("Bilangan pecahan a:", a)
print("Bilangan pecahan b:", b)
print("Menyederhanakan pecahan a:", a.limit_denominator())
print("Menyederhanakan pecahan b:", b.limit_denominator())
```
