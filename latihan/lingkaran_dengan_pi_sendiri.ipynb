# Pi (π) di Matematika SMP: 3,14 vs 22/7 📐

## 1. Apa itu Pi?

**Pi (π)** adalah perbandingan antara **keliling lingkaran** dengan **diameternya**. Rumusnya:

$$\pi = \frac{K}{d}$$

Nilai ini selalu sama untuk semua lingkaran, berapa pun ukurannya. ✨

## 2. Nilai Sebenarnya

Pi adalah **bilangan irasional**, artinya:
- Desimalnya tidak berhenti 🔢
- Desimalnya tidak berulang dengan pola tetap

$$\pi = 3{,}14159265358979...$$

Karena itu, kita memakai **nilai pendekatan** dalam perhitungan.

## 3. Perbandingan 3,14 dan 22/7

| Aspek | 3,14 | 22/7 |
|---|---|---|
| **Bentuk** | Desimal | Pecahan |
| **Nilai** | 3,14 | 3,142857... |
| **Selisih dari π** | ± 0,0016 | ± 0,0013 |
| **Lebih akurat?** | Sedikit kurang | Sedikit lebih akurat |

## 4. Kapan Memakai yang Mana? 🤔

Aturan praktis yang digunakan di SMP:

- ✅ **Pakai 22/7** jika jari-jari atau diameter merupakan **kelipatan 7** (misalnya 7, 14, 21, 28, 35).
- ✅ **Pakai 3,14** jika jari-jari atau diameter **bukan kelipatan 7** (misalnya 10, 20, 25, 30).
- 📝 Jika soal sudah menyebutkan nilai π, **ikuti soal**.

## 5. Contoh Soal

### Contoh 1: Memakai 22/7
Sebuah lingkaran memiliki jari-jari **14 cm**. Hitunglah kelilingnya!

$$K = 2 \times \pi \times r = 2 \times \frac{22}{7} \times 14 = 88 \text{ cm}$$

Angka 14 habis dibagi 7, sehingga perhitungan menjadi mudah. 👍

### Contoh 2: Memakai 3,14
Sebuah lingkaran memiliki jari-jari **10 cm**. Hitunglah luasnya!

$$L = \pi \times r^2 = 3{,}14 \times 10 \times 10 = 314 \text{ cm}^2$$

Angka 10 tidak habis dibagi 7, sehingga 3,14 lebih praktis. 👍

## 6. Rumus yang Menggunakan Pi

| Bangun | Rumus |
|---|---|
| **Keliling lingkaran** | $K = \pi \times d$ atau $K = 2\pi r$ |
| **Luas lingkaran** | $L = \pi r^2$ |
| **Luas tabung** | $L = 2\pi r(r + t)$ |
| **Volume tabung** | $V = \pi r^2 t$ |
| **Volume kerucut** | $V = \frac{1}{3}\pi r^2 t$ |
| **Volume bola** | $V = \frac{4}{3}\pi r^3$ |

## 7. Kesimpulan 📌

- **π ≈ 3,14** dan **π ≈ 22/7** hanyalah **nilai pendekatan**, bukan nilai pasti.
- **22/7** sedikit lebih dekat dengan nilai asli π.
- Pilih nilai yang membuat perhitungan **paling mudah** sesuai angka pada soal.
- Selalu perhatikan **petunjuk dalam soal**. 🎯

Semoga penjelasan ini membantu! Jika berkenan, saya dapat menyiapkan **latihan soal** beserta pembahasannya. 😊

```python
import math

def luas_lingkaran(r, pi="auto"):
    """
    Menghitung luas lingkaran.

    Parameter:
        r  : jari-jari lingkaran (harus > 0)
        pi : nilai pi yang dipakai
             - "auto" : otomatis (22/7 jika r kelipatan 7, selain itu 3,14)
             - "22/7" : paksa memakai 22/7
             - "3.14" : paksa memakai 3,14
             - "math" : memakai nilai pi presisi dari modul math
    """
    if r <= 0:
        raise ValueError("Jari-jari harus lebih besar dari 0")

    if pi == "auto":
        nilai_pi = 22 / 7 if r % 7 == 0 else 3.14
    elif pi == "22/7":
        nilai_pi = 22 / 7
    elif pi == "3.14":
        nilai_pi = 3.14
    elif pi == "math":
        nilai_pi = math.pi
    else:
        raise ValueError("Pilihan pi tidak dikenal")

    return nilai_pi * r * r

# Contoh penggunaan
print(luas_lingkaran(14))           # 616.0  (otomatis 22/7)
print(luas_lingkaran(10))           # 314.0  (otomatis 3,14)
print(luas_lingkaran(10, "math"))   # 314.1592653589793
```
