Akar Kuadrat Bilangan yang Sempurna
Bilangan yang hasil akarnya merupakan bilangan bulat disebut kuadrat sempurna.

```python
import math

print("Bilangan | Akar")
print("-" * 20)

for angka in range(1, 101):
    akar = math.sqrt(angka)

    if akar.is_integer():
        print(f"{angka:>8} | {int(akar)}")
```
