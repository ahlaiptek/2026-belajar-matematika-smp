```python
from math import gcd

pembilang = 8
penyebut = 12

fpb = gcd(pembilang, penyebut)

print("FPB:", fpb)

pembilang = pembilang // fpb
penyebut = penyebut // fpb

print("Hasil:", pembilang, "/", penyebut)
```

```python
from fractions import Fraction

a = Fraction(1, 2)
print(a)  # Output: 1/2

a = Fraction(3, 4)
print(a)  # Output: 3/4

c = Fraction(10, 5)
print(c)  # Output: 2

print()

print("Perkalian Pecahan")
d = a * c
print(f"Hasil perkalian {a} * {c} = {d}")  # Output: 3/2
```
