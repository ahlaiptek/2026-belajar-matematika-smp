```python
# Persegi
def luas_persegi(sisi):
    return sisi * sisi

luas = luas_persegi(5)
print("Luas persegi dengan sisi 5 adalah:", luas)

# Persegi Panjang
def luas_persegi_panjang(panjang, lebar):
    return panjang * lebar

luas = luas_persegi_panjang(4, 6)
print("Luas persegi panjang dengan panjang 4 dan lebar 6 adalah:", luas)
```

```python
# Lingkaran
import math

def luas_lingkaran(jari_jari):
    return math.pi * jari_jari * jari_jari

luas = luas_lingkaran(3)
print("Luas lingkaran dengan jari-jari 3 adalah:", luas)

def keliling_lingkaran(jari_jari):
    return 2 * math.pi * jari_jari

keliling = keliling_lingkaran(3)
print("Keliling lingkaran dengan jari-jari 3 adalah:", keliling)
```
