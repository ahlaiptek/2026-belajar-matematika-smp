```python
# Bilangan Bulat
"""Bilangan Bulat
- Bilangan positif dan negatif
- Membandingkan bilangan
- Penjumlahan & pengurangan
- Perkalian & pembagian
- Operasi campuran
"""

## Positif Negatif
print("Positif Negatif")
a = -5
b = -10
print(f"{a} + {b} = {a + b}")

a = -5
b = +10
print(f"{a} - {b} = {a - b}")
```

```python
## Membandingkan bilangan
a = -5
b = -10
if a > b:
    print(f"{a} lebih besar dari {b}")
elif a < b:
    print(f"{a} lebih kecil dari {b}")
else:
    print(f"{a} sama dengan {b}")
```

```python
## Nilai Mutlak
a = -5
b = -10
print(f"Nilai mutlak dari {a} adalah {abs(a)}")
print(f"Nilai mutlak dari {b} adalah {abs(b)}")
```

```python
## Operasi Campuran
"""
Ada aturan urutan operasi.
Prioritas operasi
Secara sederhana:
1. Kurung
2. Pangkat
3. Perkalian/pembagian
4. Penjumlahan/pengurangan
"""
a = 5 + 3 * 2
print(f"5 + 3 * 2 = {a}")
```
